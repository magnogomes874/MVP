```python
import sqlite3
import pandas as pd
import ipywidgets as widgets
from IPython.display import display, clear_output, HTML

# LISTA DE PRODUTOS
# Estrutura: (Código, Descrição, Categoria, Quantidade, Estoque Mínimo, Custo Unitário)

DADOS_INICIAIS = [
    ("MAT-001", "Parafuso", "Fixadores", 450.0, 100.0, 0.45),
    ("MAT-002", "Chapa de Aço", "Matéria-Prima", 18.0, 25.0, 120.00),
    ("MAT-003", "Rolamento", "Componentes", 8.0, 15.0, 32.50),
    ("MAT-004", "Sensor Indutivo", "Eletrônica", 35.0, 10.0, 85.00),
    ("MAT-005", "Tinta Primer Cinza", "Químicos", 12.0, 8.0, 95.00),
    ("MAT-006", "Cabo Flexível", "Elétrica", 220.0, 150.0, 3.20)
]

CATEGORIAS = ["Matéria-Prima", "Componentes", "Fixadores", "Elétrica", "Eletrônica", "Químicos", "Outros"]

# 1. BANCO DE DADOS
def init_db():
    conn = sqlite3.connect("estoque_colab.db")
    cursor = conn.cursor()
    cursor.execute("""
        CREATE TABLE IF NOT EXISTS estoque (
            id INTEGER PRIMARY KEY AUTOINCREMENT,
            codigo TEXT UNIQUE,
            descricao TEXT,
            categoria TEXT,
            quantidade REAL,
            estoque_minimo REAL,
            custo_unitario REAL
        )
    """)

    # Atualiza ou insere a lista do código sem travar o banco
    for item in DADOS_INICIAIS:
        cursor.execute("""
            INSERT INTO estoque (codigo, descricao, categoria, quantidade, estoque_minimo, custo_unitario)
            VALUES (?, ?, ?, ?, ?, ?)
            ON CONFLICT(codigo) DO UPDATE SET
                descricao = excluded.descricao,
                categoria = excluded.categoria,
                estoque_minimo = excluded.estoque_minimo,
                custo_unitario = excluded.custo_unitario
        """, item)

    conn.commit()
    conn.close()

init_db()

def get_df():
    conn = sqlite3.connect("estoque_colab.db")
    df = pd.read_sql("SELECT * FROM estoque", conn)
    conn.close()
    return df

# 2. ÁREAS DE EXIBIÇÃO
out_kpis = widgets.Output()
out_tabela = widgets.Output()
out_msg = widgets.Output()

#3.WIDGETS DE OPERAÇÃO

#Aba:Movimentação
dropdown_itens = widgets.Dropdown(description="Item:", layout=widgets.Layout(width='60%'))
tipo_mov = widgets.RadioButtons(options=['Entrada (+)', 'Saída (-)'], description='Tipo:')
qtd_mov = widgets.FloatText(value=1.0, description='Qtd:', layout=widgets.Layout(width='30%'))
btn_mov = widgets.Button(description="Registrar Movimentação", button_style='primary')

def registrar_movimentacao(b):
    if not dropdown_itens.value:
        return
    cod = dropdown_itens.value
    qtd = qtd_mov.value
    df = get_df()
    atual = df.loc[df["codigo"] == cod, "quantidade"].values[0]
    nova = (atual + qtd) if tipo_mov.value == 'Entrada (+)' else (atual - qtd)

    with out_msg:
        clear_output()
        if nova < 0:
            display(HTML("<span style='color:red;'>❌ Erro: Quantidade não pode ficar negativa.</span>"))
            return
        conn = sqlite3.connect("estoque_colab.db")
        conn.execute("UPDATE estoque SET quantidade = ? WHERE codigo = ?", (nova, cod))
        conn.commit()
        conn.close()
        display(HTML("<span style='color:green;'>✔️ Movimentação registrada!</span>"))
    atualizar_interface()

btn_mov.on_click(registrar_movimentacao)

# Aba:Novo Cadastro
txt_cod = widgets.Text(description="Código SKU:")
txt_desc = widgets.Text(description="Descrição:")
drop_cat = widgets.Dropdown(options=CATEGORIAS, description="Categoria:")
num_qtd = widgets.FloatText(value=0.0, description="Estoque Inicial:")
num_min = widgets.FloatText(value=0.0, description="Estoque Mín:")
num_custo = widgets.FloatText(value=0.0, description="Custo Unit:")
btn_cad = widgets.Button(description="Cadastrar Item", button_style='success')

def registrar_cadastro(b):
    with out_msg:
        clear_output()
        if not txt_cod.value or not txt_desc.value:
            display(HTML("<span style='color:orange;'>⚠️ Preencha ao menos Código e Descrição.</span>"))
            return
        try:
            conn = sqlite3.connect("estoque_colab.db")
            conn.execute("""
                INSERT INTO estoque (codigo, descricao, categoria, quantidade, estoque_minimo, custo_unitario)
                VALUES (?, ?, ?, ?, ?, ?)
            """, (txt_cod.value.strip().upper(), txt_desc.value.strip(), drop_cat.value, num_qtd.value, num_min.value, num_custo.value))
            conn.commit()
            conn.close()
            display(HTML("<span style='color:green;'>✔️ Item cadastrado!</span>"))
            txt_cod.value = ""
            txt_desc.value = ""
            atualizar_interface()
        except sqlite3.IntegrityError:
            display(HTML("<span style='color:red;'>❌ Erro: SKU já existe.</span>"))

btn_cad.on_click(registrar_cadastro)

# Aba:Editar/Excluir
dropdown_edit = widgets.Dropdown(description="Item:", layout=widgets.Layout(width='60%'))
edit_desc = widgets.Text(description="Descrição:")
edit_cat = widgets.Dropdown(options=CATEGORIAS, description="Categoria:")
edit_min = widgets.FloatText(description="Estoque Mín:")
edit_custo = widgets.FloatText(description="Custo Unit:")
btn_salvar_edit = widgets.Button(description="Salvar Alterações", button_style='info')
btn_excluir = widgets.Button(description="Excluir Item", button_style='danger')

def carregar_dados_edicao(change):
    if dropdown_edit.value:
        df = get_df()
        item = df[df["codigo"] == dropdown_edit.value]
        if not item.empty:
            edit_desc.value = str(item.iloc[0]["descricao"])
            cat_atual = str(item.iloc[0]["categoria"])
            edit_cat.value = cat_atual if cat_atual in CATEGORIAS else CATEGORIAS[0]
            edit_min.value = float(item.iloc[0]["estoque_minimo"])
            edit_custo.value = float(item.iloc[0]["custo_unitario"])

dropdown_edit.observe(carregar_dados_edicao, names='value')

def salvar_edicao(b):
    cod = dropdown_edit.value
    if not cod:
        return
    with out_msg:
        clear_output()
        conn = sqlite3.connect("estoque_colab.db")
        conn.execute("""
            UPDATE estoque
            SET descricao = ?, categoria = ?, estoque_minimo = ?, custo_unitario = ?
            WHERE codigo = ?
        """, (edit_desc.value.strip(), edit_cat.value, edit_min.value, edit_custo.value, cod))
        conn.commit()
        conn.close()
        display(HTML("<span style='color:green;'>✔️ Alterações salvas!</span>"))
    atualizar_interface()

btn_salvar_edit.on_click(salvar_edicao)

def excluir_item(b):
    cod = dropdown_edit.value
    if not cod:
        return
    with out_msg:
        clear_output()
        conn = sqlite3.connect("estoque_colab.db")
        conn.execute("DELETE FROM estoque WHERE codigo = ?", (cod,))
        conn.commit()
        conn.close()
        display(HTML(f"<span style='color:red;'>🗑️ Item {cod} excluído!</span>"))
    atualizar_interface()

btn_excluir.on_click(excluir_item)

# 4.VISUAL
def atualizar_interface():
    df = get_df()
    df["valor_total"] = df["quantidade"] * df["custo_unitario"]

    total_skus = len(df)
    valor_total = df["valor_total"].sum()
    criticos = df[df["quantidade"] <= df["estoque_minimo"]]

    with out_kpis:
        clear_output()
        display(HTML(f"""
            <div style="display: flex; gap: 15px; margin-bottom: 15px; font-family: sans-serif;">
                <div style="background: #f0f2f6; padding: 10px 15px; border-radius: 8px; flex: 1;">
                    <small>Total de SKUs</small><h3>{total_skus}</h3>
                </div>
                <div style="background: #f0f2f6; padding: 10px 15px; border-radius: 8px; flex: 1;">
                    <small>Valor em Estoque</small><h3>R$ {valor_total:,.2f}</h3>
                </div>
                <div style="background: {'#ffebe6' if len(criticos) > 0 else '#f0f2f6'}; padding: 10px 15px; border-radius: 8px; flex: 1;">
                    <small>Itens Críticos</small><h3 style="color: {'#d9381e' if len(criticos) > 0 else '#333'};">{len(criticos)}</h3>
                </div>
            </div>
        """))

    with out_tabela:
        clear_output()
        if len(criticos) > 0:
            display(HTML(f"<b style='color:red;'>⚠️ Atenção: {len(criticos)} item(ns) no limite ou abaixo do estoque mínimo!</b>"))

        display(df[["codigo", "descricao", "categoria", "quantidade", "estoque_minimo", "custo_unitario", "valor_total"]].style.format({
            "quantidade": "{:.1f}",
            "estoque_minimo": "{:.1f}",
            "custo_unitario": "R$ {:.2f}",
            "valor_total": "R$ {:.2f}"
        }))

    opcoes = [(f"{r['codigo']} - {r['descricao']}", r['codigo']) for _, r in df.iterrows()]
    dropdown_itens.options = opcoes
    dropdown_edit.options = opcoes
    if opcoes and not dropdown_edit.value:
        dropdown_edit.value = opcoes[0][1]
        carregar_dados_edicao(None)

# 5 MONTAGEM DA TELA
form_mov = widgets.VBox([dropdown_itens, tipo_mov, qtd_mov, btn_mov])
form_cad = widgets.VBox([txt_cod, txt_desc, drop_cat, num_qtd, num_min, num_custo, btn_cad])
form_edit = widgets.VBox([dropdown_edit, edit_desc, edit_cat, edit_min, edit_custo, widgets.HBox([btn_salvar_edit, btn_excluir])])

abas = widgets.Tab(children=[form_mov, form_cad, form_edit])
abas.set_title(0, "📦 Movimentação")
abas.set_title(1, "➕ Novo Cadastro")
abas.set_title(2, "✏️ Editar / Excluir")

display(HTML("<h2>📦 Painel de Controle de Estoque </h2>"))
display(out_kpis)
display(out_tabela)
display(HTML("<hr><h3>Operações</h3>"))
display(abas)
display(out_msg)

atualizar_interface()
```
